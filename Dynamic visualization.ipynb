### Скользящее среднее

Это не пункт из создания графика, но важное знание при работе с временными рядами.  

**[Скользящее среднее (MA)](https://ru.wikipedia.org/wiki/%D0%A1%D0%BA%D0%BE%D0%BB%D1%8C%D0%B7%D1%8F%D1%89%D0%B0%D1%8F_%D1%81%D1%80%D0%B5%D0%B4%D0%BD%D1%8F%D1%8F)** — усреднение значений за определённый период, сдвигающийся во времени. Оно нужно для того, чтобы например отсматривать сигналы (для финансов), **определять тренд** как краткосрочный, так и долгосрочный.

**Простое скользящее среднее** считается так
$$SMA_t = \frac{1}{n} \sum\limits_{i=0}^{n-1} p_{t-1},$$
то есть как среднее по окну.

Скользящее среднее можно считать по разному, но мы в нашем задании ограничимся простым. Для подсчета его в пандасе как раз есть [штука](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.rolling.html) `.rolling`, которая вам поможет.

# Задание 2: сезонность (2 pt)

#### Необходимая информация

**Сезонность** — это повторяющиеся паттерны во временном ряду, которые возникают через фиксированные промежутки времени. Само по себе слово общее, под ним могут понимать разное, поэтому давайте разделим.
***

* **Тренд** — долгосрочное направление движения значений.
* **Сезонность** — периодические колебания в определенное время (например перед праздниками). Может быть
  - _**дневная**_ (привязана к часам),
  - _**недельная**_ (призана к дням недели),
  - _**месячная**_ (может быть привязана к отчетному периоду или времени выплат),
  - _**годовая**_ (привязана к месяцам и важным событиям).
* **Цикл** — долгосрочные колебания без фиксированного периода
* **Остаток** — случайные колебания (шум).

Тогда значения во временном ряду можно **декомпозировать**, например, как сумму тренда, сезонности и остатка.

Мы не будем подробно разбирать декомпозицию, тут нам достаточно знать, что есть замечательная библиотека `statsmodels`, в которой реализованы методы, которые умеют выделять сезонность:
- [классическая](https://www.statsmodels.org/stable/generated/statsmodels.tsa.seasonal.seasonal_decompose.html) декомпозиция в `seasonal_decompose`,
- [более умная](https://www.statsmodels.org/dev/generated/statsmodels.tsa.seasonal.STL.html) декомпозиция в `STL`.

Не трудно догадаться, что фича «тикер» является **категориальной**, и мы можем это использовать в нашей визуализации.

Начнем немного издалека и поговорим про **cross-filtering**.  
**Cross-filtering** — это техника, при которой взаимодействие с одним графиком влияет на отображение других графиков.

**Типы связей между графиками**

* **Shared Axes** — общая ось X или Y (zoom на одном → zoom на другом),
* **Legend Toggle** — клик на легенде → показать/скрыть (включить/выключить тикер),
* **Brushing** — выделение области → фильтр других (выбрать период → обновить статистику),
* **Click Events** — клик на точке → действие (клик на акции → показать детали),
* **Dropdown** — выбор в меню → обновление (выбрать метрику → изменить график).

То есть вы можете сделать так, чтобы на графике можно было **выбирать, какие компании показывать**: можно посмотреть все, можно отключить некоторые, можно оставить только одну. И более того, вы можете сделать так, чтобы выбор компании **влиял сразу на все подграфики** (обновлял масштаб, внешний вид, значения)!

## Задание (4 pt)

Вам необходимо самостоятельно решить, какую именно информацию о компаниях вы хотите отобразить на графиках, и построить **несколько подграфиков на одной фигуре, связанных между собой**, а также сделать по ним **ВЫВОДЫ**.

Выбор фич полностью остается за вами, однако графики должны соответствовать следующим **трем критериям**:
* Должны быть **кнопки/фильтры выбора компаний**, которые влияют на отображение.
* Должно быть **минимум два связанных графика**, где изменение одного влияет на изменение другого.
* Должны быть минимум **три графика**, отображающие **разную информацию** (разные цены разной информацией не считаются).

Конечно, все это должно сопровождаться **легендой**, **названиями осей** и **hover**.

_Каждый пункт и выводы оцениваются в 1 pt._

````python
# Подготовка данных для сравнения всех тикеров
df_compare = df.sort_values(['Ticker', 'Date']).copy()
tickers = sorted(df_compare['Ticker'].unique())
groups = df_compare.groupby('Ticker')

# Начальная цена каждого тикера = 100
df_compare['PriceIndex'] = (
    df_compare['Close'] / groups['Close'].transform('first') * 100
)
df_compare['Return'] = groups['Close'].pct_change(fill_method=None)

# Годовая волатильность по окну 20 торговых сессий, 251 сессия в году
df_compare['Volatility20'] = (
    df_compare.groupby('Ticker')['Return']
    .transform(lambda x: x.rolling(20).std()) * np.sqrt(251) * 100
)

# Объём относительно своего среднего за 20 сессий (включая текущую)
df_compare['RelativeVolume'] = (
    df_compare['Volume'] / groups['Volume'].transform(lambda x: x.rolling(20).mean())
)
````

````python
# Три связанных графика с общей временной осью
fig = make_subplots(
    rows=3, cols=1, shared_xaxes=True,
    vertical_spacing=0.09,
    subplot_titles=(
        'Динамика цены: первая торговая сессия = 100',
        'Годовая волатильность: окно 20 торговых сессий',
        'Объём торгов относительно среднего за 20 сессий'
    )
)

colors = ['royalblue', 'darkorange', 'seagreen', 'firebrick', 'mediumpurple', 'slategray']
line_styles = ['solid', 'dash', 'dot', 'dashdot', 'longdash', 'longdashdot']
metrics = [
    ('PriceIndex', 'Индекс цены', ''),
    ('Volatility20', 'Волатильность', '%'),
    ('RelativeVolume', 'Относительный объём', '×')
]

for ticker, color, line_style in zip(tickers, colors, line_styles):
    data = df_compare[df_compare['Ticker'] == ticker]
    for row, (metric, label, unit) in enumerate(metrics, start=1):
        fig.add_trace(
            go.Scatter(
                x=data['Date'], y=data[metric], mode='lines',
                name=ticker, legendgroup=ticker, showlegend=(row == 1),
                line={'color': color, 'dash': line_style, 'width': 1.5},
                connectgaps=False,
                hovertemplate=(
                    f'{ticker}<br>Дата: %{{x|%Y-%m-%d}}<br>'
                    f'{label}: %{{y:.2f}}{unit}<extra></extra>'
                )
            ), row=row, col=1
        )

# Меню меняет видимость тикера одновременно на всех трёх графиках
buttons = [{
    'label': 'Все тикеры', 'method': 'update',
    'args': [{'visible': [True] * len(fig.data)}]
}]
for ticker in tickers:
    buttons.append({
        'label': ticker, 'method': 'update',
        'args': [{'visible': [trace.legendgroup == ticker for trace in fig.data]}]
    })

fig.update_layout(
    title='Сравнение акций и индекса S&P 500, 2021–2024',
    height=1150, template='plotly_white',
    hovermode='x unified', dragmode='zoom',
    legend={
        'title': 'Тикер (клик — скрыть/показать на всех графиках)',
        'orientation': 'h', 'x': 0, 'y': 1.11,
        'groupclick': 'togglegroup'
    },
    updatemenus=[{
        'buttons': buttons, 'direction': 'down',
        'x': 1, 'xanchor': 'right', 'y': 1.21, 'yanchor': 'top', 'showactive': True
    }],
    margin={'l': 100, 'r': 50, 't': 220, 'b': 70}
)

# shared_xaxes связывает zoom/pan: изменение периода обновляет все панели
fig.update_xaxes(tickformat='%Y-%m', showticklabels=True)
fig.update_xaxes(title_text='Дата', row=3, col=1)
fig.update_yaxes(title_text='Индекс цены', row=1, col=1)
fig.update_yaxes(title_text='Волатильность, %', rangemode='tozero', row=2, col=1)
fig.update_yaxes(title_text='Объём / среднее, ×', rangemode='tozero', row=3, col=1)
fig.add_hline(y=100, line_dash='dot', line_color='gray', row=1, col=1)
fig.add_hline(y=1, line_dash='dot', line_color='gray', row=3, col=1)

Path('graphs').mkdir(exist_ok=True)
config = {'scrollZoom': True, 'displaylogo': False}
fig.write_html('graphs/task_3.html', include_plotlyjs=True, full_html=True, config=config)
fig.write_image('graphs/task_3.png', width=1400, height=1150, scale=1)
fig.show(config=config)
````

Эта библиотека позволяет делать **дашборды**, но **требует более детальной настройки** интерактивности (а также не позволяет работать на python без сервера со многими штуками, поэтому не пугайтесь js).

Задание заключается в том, чтобы **дополнить код** для получения графика.

На графике должны быть отображены **цена закрытия** и **скользящее среднее** для акций одной компании. Должна присутствовать **возможность отключать графики**, должен быть **hover**.

Шаблон есть, пропуски довольно очевидны. если вы хотите поразбираться в библиотеке, то можете сделать _помимо этого_ более сложные графики (например повторить предыдущие задания) на дополнительные баллы.

````python
from bokeh.layouts import column
from bokeh.models import ColumnDataSource, CustomJS, HoverTool, Select
from bokeh.plotting import figure, output_file, save
from bokeh.resources import INLINE

TICKER = "AAPL"

# Срез по последним сессиям и подсчёт скользящего среднего
def prepare_data(ticker, days=252):
    df_ticker = df[df['Ticker'] == ticker].copy()
    df_ticker = df_ticker.sort_values('Date')
    # Считаем до среза, чтобы использовать историю перед началом окна
    df_ticker['MA20'] = df_ticker['Close'].rolling(window=20).mean()
    return df_ticker.tail(days)

df_initial = prepare_data(TICKER)

source = ColumnDataSource(data={
    'Date': df_initial['Date'],
    'Close': df_initial['Close'],
    'MA20': df_initial['MA20'],
    'Volume': df_initial['Volume']
})
````

````python
# Создание основной фигуры
p = figure(
    title=f"{TICKER} — цена закрытия и MA20, последние 252 торговые сессии",
    x_axis_type='datetime',
    height=550,
    width=1100,
    tools="pan,wheel_zoom,box_zoom,reset,save",
    toolbar_location="above",
    sizing_mode='stretch_width'
)

# Линия цены
price_line = p.line(
    x='Date', y='Close',
    source=source,
    line_width=2,
    color='royalblue',
    legend_label='Цена закрытия'
)

# Скользящее среднее
ma_line = p.line(
    x='Date', y='MA20',
    source=source,
    line_width=2,
    color='darkorange',
    line_dash='dashed',
    legend_label='MA20'
)

# Hover с данными обеих линий
p.add_tools(HoverTool(
    renderers=[price_line, ma_line],
    tooltips=[
        ("Дата", "@Date{%Y-%m-%d}"),
        ("Цена закрытия", "@Close{0,0.00}"),
        ("MA20", "@MA20{0,0.00}"),
        ("Объём торгов", "@Volume{0,0}")
    ],
    formatters={'@Date': 'datetime'},
    mode='vline',
    muted_policy='ignore'
))

# Настройка легенды и осей
p.legend.location = "top_left"
p.legend.click_policy = "hide"
p.xaxis.axis_label = "Дата"
p.yaxis.axis_label = "Цена, USD"
p.grid.grid_line_alpha = 0.25

# Dropdown для выбора тикера
ticker_select = Select(
    title="Выберите тикер (^GSPC — индекс S&P 500):",
    value=TICKER,
    options=sorted(df['Ticker'].unique().tolist()),
    width=350
)

# Данные всех тикеров для автономного HTML
all_data = {}
for ticker in ticker_select.options:
    df_ticker = prepare_data(ticker)
    all_data[ticker] = {
        'Date': df_ticker['Date'].to_numpy(dtype='datetime64[ms]').astype('int64').tolist(),
        'Close': df_ticker['Close'].tolist(),
        'MA20': df_ticker['MA20'].to_numpy(),
        'Volume': df_ticker['Volume'].tolist()
    }

# JavaScript callback обновляет обе линии без Python-сервера
callback = CustomJS(args={
    'source': source, 'data': all_data, 'title_obj': p.title, 'y_axis': p.yaxis[0]
}, code="""
    const ticker = cb_obj.value;
    source.data = data[ticker];
    source.change.emit();
    title_obj.text = ticker + ' — цена закрытия и MA20, последние 252 торговые сессии';
    y_axis.axis_label = ticker === '^GSPC' ? 'Индекс, пункты' : 'Цена, USD';
""")

ticker_select.js_on_change('value', callback)

# Компоновка и сохранение вместе с библиотекой Bokeh для работы без интернета
layout = column(ticker_select, p, sizing_mode='stretch_width', margin=20)
Path('graphs').mkdir(exist_ok=True)
output_file("graphs/task_4.html", title="Цена закрытия и MA20 - Bokeh")
save(layout, resources=INLINE)
````

_Что удобнее: plotly или bokeh?_

Для этих заданий мне был удобен Plotly, потому что в нём проще создавать подграфики, настраивать общие оси и фильтры. В Bokeh нужно подробнее описывать источники данных и обработчики, зато связь виджетов с графиком получается явной.

Здесь выбор тикера обновляет цену закрытия и MA20 через общий ColumnDataSource. Линии можно независимо отключать кликом по легенде, а при наведении доступны дата, цена, MA20 и объём торгов. Для GSPC ось отображает пункты индекса.
